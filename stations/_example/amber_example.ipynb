## 3. The data folder (optional)

If you have the run's data folder (made by AMBER on the instrument PC, D-73), point to it. `AMBER_run.json` holds the run, the experiment and the samples.

```python
from pathlib import Path

FOLDER = Path("data") / RUN   # change me: where the run's folder is

if (FOLDER / "AMBER_run.json").exists():
    info = load_run_json(FOLDER)
    samples = info["samples"]
    print(info["run"]["code"], "·", info["run"]["title"])
    for s in samples:
        print(" ", s["code"], s.get("title") or "", s.get("composition") or "")
else:
    print("No data folder at", FOLDER, "- upload it or change FOLDER")
```

## 4. A first plot

Reads the first CSV file in `processed/` (or `raw/`) and plots its first two columns. Replace it with your station's analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt

files = sorted((FOLDER / "processed").glob("*.csv")) or sorted((FOLDER / "raw").glob("*.csv"))
if files:
    df = pd.read_csv(files[0], sep=None, engine="python")
    x, y = df.columns[:2]
    ax = df.plot(x=x, y=y, legend=False)
    ax.set_xlabel(x); ax.set_ylabel(y); ax.set_title(f"{RUN} · {files[0].name}")
    plt.show()
else:
    print("No CSV files yet")
```
